```python
import pandas as pd
```

```python
df = pd.read_parquet("synthetic_dataset_100k.pq")
# remove all rows where the length of the prompt is greater than 1000
df = df[df["prompt"].apply(len) <= 1000]
```

```python
df_unique = df.drop_duplicates(subset=['prompt'], keep='first')
df_unique = df_unique.reset_index(drop=True)
df_unique
```

```python
df_unique.to_parquet("poison_dataset_100k_final.pq")
```
